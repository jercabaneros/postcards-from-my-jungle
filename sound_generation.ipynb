```python
!pip install torchsde
import torch
import soundfile as sf
from diffusers import StableAudioPipeline
from IPython.display import Audio
import os

# load up the model
pipe = StableAudioPipeline.from_pretrained("stabilityai/stable-audio-open-1.0", torch_dtype=torch.float16)
pipe = pipe.to("cuda")

os.makedirs("sounds", exist_ok=True)

# my 4 different bird sound prompts
prompts = [
    {
        "text": "Loud zoo aviary sounds, multiple birds calling at once, squawking parrots, chattering macaws, cranes calling, ibis screeching, constant movement, flapping wings, rustling feathers, echoing enclosure walls, energetic zoo soundscape",
        "negative": "music, melody, human voices, footsteps, talking, singing, announcer, ambient peaceful atmosphere",
        "duration": 12,
        "filename": "zoo_aviary_mixed.wav"
    },
    {
        "text": "Lively zoo mammal enclosure, monkeys chattering loudly, lemurs calling, small mammals scurrying, rustling leaves, energetic movement, distant roars and growls from large cats, enclosure echo, chaotic and busy animal activity",
        "negative": "music, humans, speech, footsteps, calm ambience, quiet forest atmosphere",
        "duration": 12,
        "filename": "zoo_mammal_enclosure.wav"
    },

    {
        "text": "Busy zoo aviary, many birds calling loudly, parrots shrieking, cockatoos squawking, small birds chirping rapidly, wings fluttering, cage echoes, energetic and noisy, chaotic bird environment with constant animal activity",
        "negative": "music, human voices, calm ambience, soft nature recordings",
        "duration": 15,
        "filename": "zoo_aviary_morning.wav"
    },

    {
        "text": "Energetic zoo bird enclosure, dominance calls, fast beak clacks, rapid wing flaps, birds hopping and pacing, sharp squawks and bursts of calls, lively activity echoing off enclosure surfaces",
        "negative": "violence, humans, music, calm ambience, peaceful nature atmosphere",
        "duration": 12,
        "filename": "zoo_bird_activity.wav"
    }
]

# generate each sound
for i, prompt_info in enumerate(prompts):
    generator = torch.Generator("cuda").manual_seed(42 + i)

    audio = pipe(
        prompt_info["text"],
        negative_prompt=prompt_info["negative"],
        num_inference_steps=200,
        audio_end_in_s=prompt_info["duration"],
        num_waveforms_per_prompt=1,
        generator=generator,
    ).audios

    # save it
    output_path = f"sounds/{prompt_info['filename']}"
    output = audio[0].T.float().cpu().numpy()
    sf.write(output_path, output, pipe.vae.sampling_rate)

    # listen to it
    display(Audio(output_path))

    torch.cuda.empty_cache()

# zip everything up
import shutil
shutil.make_archive("sounds", "zip", "sounds")

# download in colab
try:
    from google.colab import files
    files.download('sounds.zip')
except:
    pass
```
